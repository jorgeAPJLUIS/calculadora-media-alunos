```python
# Programa para calcular a média das notas de um aluno

# Função que calcula a média das notas
def calcular_media(notas):
    return sum(notas) / len(notas)


# Lista que irá armazenar as notas
notas = []

# Pergunta quantas notas serão cadastradas
quantidade = int(input("Quantas notas deseja informar? "))

# Repetição para receber as notas
for i in range(quantidade):
    nota = float(input(f"Digite a {i + 1}ª nota: "))
    notas.append(nota)

# Chama a função para calcular a média
media = calcular_media(notas)

# Mostra as notas informadas
print("\nNotas informadas:", notas)

# Mostra a média calculada
print(f"Média final: {media:.2f}")

# Verifica se o aluno foi aprovado ou reprovado
if media >= 7:
    print("Resultado: Aprovado")
else:
    print("Resultado: Reprovado")
```
